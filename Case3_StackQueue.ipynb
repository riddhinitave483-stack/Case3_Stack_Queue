```python

# Case Study 3: Hospital Patient Management
# Stack and Queue using Linked Lists

class Node:
    def __init__(self, name):
        self.name = name
        self.next = None


# Queue: Waiting Patients (FIFO)
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self, name):
        new_node = Node(name)

        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        print(name, "added to waiting queue.")

    def dequeue(self):
        if self.front is None:
            print("No patients waiting.")
            return

        name = self.front.name
        self.front = self.front.next

        if self.front is None:
            self.rear = None

        print("Now treating patient:", name)

    def display(self):
        if self.front is None:
            print("No waiting patients.")
            return

        current = self.front
        print("\nWaiting Patients:")

        while current:
            print("->", current.name)
            current = current.next


# Stack: Registration History (LIFO)
class Stack:
    def __init__(self):
        self.top = None

    def push(self, name):
        new_node = Node(name)
        new_node.next = self.top
        self.top = new_node
        print(name, "added to registration history.")

    def pop(self):
        if self.top is None:
            print("Registration history is empty.")
            return

        print("Removed from history:", self.top.name)
        self.top = self.top.next

    def display(self):
        if self.top is None:
            print("Registration history is empty.")
            return

        current = self.top
        print("\nRegistration History (latest first):")

        while current:
            print("->", current.name)
            current = current.next


# Main Program
queue = Queue()
stack = Stack()

while True:
    print("\n===== Hospital Patient Management =====")
    print("1. Register Patient")
    print("2. Call Next Patient")
    print("3. Show Waiting Patients")
    print("4. Show Registration History")
    print("5. Remove Last History Entry")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ").strip()

    if choice == "1":
        name = input("Enter patient name: ").strip()

        if name:
            queue.enqueue(name)
            stack.push(name)
        else:
            print("Patient name cannot be empty.")

    elif choice == "2":
        queue.dequeue()

    elif choice == "3":
        queue.display()

    elif choice == "4":
        stack.display()

    elif choice == "5":
        stack.pop()

    elif choice == "6":
        print("Hospital Management System Closed.")
        break

    else:
        print("Invalid choice! Please enter 1 to 6.")
```
